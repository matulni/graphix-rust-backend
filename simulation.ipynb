```python
import numpy as np
from graphix.clifford import Clifford
from graphix.command import C, Z
from graphix.pattern import Pattern
from graphix.states import BasicStates

# Correct behavior
p = Pattern(input_nodes=[0], cmds=[C(0, Clifford.Z)])
state = p.simulate_pattern(input_state=BasicStates.PLUS)
print(state)  # [ 0.70710678+0.j -0.70710678+0.j]
assert np.isclose(np.abs(np.dot(state.flatten().conjugate(), BasicStates.MINUS.get_statevector())), 1)

# Incorrect behavior
p = Pattern(input_nodes=[0], cmds=[Z(0)])
state = p.simulate_pattern(input_state=BasicStates.PLUS)
print(state)  # [ 0.70710678+0.j 0.70710678+0.j]
assert np.isclose(np.abs(np.dot(state.flatten().conjugate(), BasicStates.MINUS.get_statevector())), 1)
```
